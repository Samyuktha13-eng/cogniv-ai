```python
class AdaptationAgent:

    levels = [
        "very_easy",
        "easy",
        "medium",
        "hard"
    ]

    def update(
        self,
        difficulty,
        correct,
        hints_used,
        response_time,
        independent
    ):

        if difficulty not in self.levels:
            raise ValueError(
                f"Unknown difficulty level: {difficulty}"
            )

        index = self.levels.index(difficulty)

        # Strong independent performance
        if (
            correct
            and independent
            and hints_used == 0
            and response_time < 10
        ):
            index += 1

        # Struggling performance
        elif (
            not correct
            or hints_used >= 2
            or response_time > 30
        ):
            index -= 1

        # Keep difficulty within valid range
        index = max(
            0,
            min(index, len(self.levels) - 1)
        )

        return self.levels[index]
```

```python
adaptation_agent = AdaptationAgent()

print("Adaptation Agent ready!")
```

```python
next_difficulty = adaptation_agent.update(
    difficulty="very_easy",
    correct=True,
    hints_used=0,
    response_time=5,
    independent=True
)

print("Next difficulty:", next_difficulty)
```

```python
next_difficulty = adaptation_agent.update(
    difficulty="medium",
    correct=False,
    hints_used=2,
    response_time=40,
    independent=False
)

print("Next difficulty:", next_difficulty)
```
