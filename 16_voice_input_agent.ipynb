```python
from pathlib import Path
import os
import time

from dotenv import load_dotenv
from huggingface_hub import login
from transformers import AutoModel

PROJECT_DIR = Path(r"C:\Users\dmane\Downloads\Cogniv AI")
ENV_FILE = PROJECT_DIR / ".env"

# Load the project's .env file explicitly
load_dotenv(ENV_FILE, override=True)

HF_TOKEN = os.getenv("HF_TOKEN")

print("Environment file:", ENV_FILE)
print(".env exists:", ENV_FILE.exists())
print("HF_TOKEN loaded:", bool(HF_TOKEN))

if not HF_TOKEN:
    raise RuntimeError(
        f"HF_TOKEN was not loaded from {ENV_FILE}. "
        "Check that the .env file contains HF_TOKEN=..."
    )

# Authenticate with Hugging Face
login(token=HF_TOKEN, add_to_git_credential=False)

print("Hugging Face authentication successful.")
print("Loading IndicConformer...")

load_start = time.perf_counter()

model = AutoModel.from_pretrained(
    MODEL_ID,
    trust_remote_code=True,
    token=HF_TOKEN,
)

load_time = time.perf_counter() - load_start

print(f"\nModel loaded in {load_time:.2f} seconds")
print("Model type:", type(model))
```
