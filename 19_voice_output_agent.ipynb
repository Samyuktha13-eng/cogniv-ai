```python
import torch

print("Loading Svara SNAC decoder...")

snac_codec = SNACCodec(device="cuda")

print("\nSNAC decoder loaded successfully.")
print("Device:", snac_codec.device)
print("Sample rate:", snac_codec.sample_rate)
```

```python
TEST_TEXT = "Hello, how are you today?"

speaker_id = "Indian English (Female)"

prompt_token_ids = svara_text_to_tokens(
    text=TEST_TEXT,
    speaker_id=speaker_id,
    tokenizer=tokenizer,
)

print("Svara prompt created successfully.")
print("Number of prompt tokens:", len(prompt_token_ids))
print("First 20 token IDs:", prompt_token_ids[:20])
print("Last 10 token IDs:", prompt_token_ids[-10:])
```

```python
print("Generating Svara speech tokens...")

input_ids = torch.tensor(
    [prompt_token_ids],
    dtype=torch.long,
    device=model.get_input_embeddings().weight.device,
)

with torch.inference_mode():
    generated = model.generate(
        input_ids=input_ids,
        max_new_tokens=400,
        do_sample=True,
        temperature=0.7,
        top_p=0.9,
    )

print("Generation complete.")
print("Generated shape:", generated.shape)

new_tokens = generated[0, len(prompt_token_ids):].tolist()

print("New tokens generated:", len(new_tokens))
print("First 30 generated tokens:")
print(new_tokens[:30])
```

```python
print("Regenerating Svara speech with EOS stopping...")

with torch.inference_mode():
    generated = model.generate(
        input_ids=input_ids,
        max_new_tokens=1200,
        do_sample=True,
        temperature=0.7,
        top_p=0.9,
        repetition_penalty=1.0,
        eos_token_id=128258,
    )

print("Generation complete.")
print("Generated shape:", generated.shape)

new_tokens = generated[0, len(prompt_token_ids):].tolist()

print("New tokens generated:", len(new_tokens))
print("First 30 generated tokens:")
print(new_tokens[:30])

print("Last 10 generated tokens:")
print(new_tokens[-10:])

print("EOS present:", 128258 in new_tokens)
```

```python
import torch
import numpy as np
import wave
from IPython.display import Audio, display

print("Converting Svara audio tokens to SNAC codes...")

# ---------------------------------------------------------
# 1. Keep only Svara audio tokens
# ---------------------------------------------------------
AUDIO_TOKENS_START = 128266
AUDIO_VOCAB_SIZE = 4096

audio_tokens = [
    int(t)
    for t in new_tokens
    if AUDIO_TOKENS_START <= int(t) < AUDIO_TOKENS_START + 7 * AUDIO_VOCAB_SIZE
]

# Keep only complete 7-token frames
audio_tokens = audio_tokens[:(len(audio_tokens) // 7) * 7]

print("Audio tokens:", len(audio_tokens))
print("Complete 7-band frames:", len(audio_tokens) // 7)

# ---------------------------------------------------------
# 2. Convert Svara token IDs -> SNAC codes
# ---------------------------------------------------------
codes_0 = []
codes_1 = []
codes_2 = []

for i in range(0, len(audio_tokens), 7):
    frame = audio_tokens[i:i + 7]

    codes_0.append(
        frame[0] - AUDIO_TOKENS_START
    )

    codes_1.extend([
        frame[1] - AUDIO_TOKENS_START - 1 * AUDIO_VOCAB_SIZE,
        frame[4] - AUDIO_TOKENS_START - 4 * AUDIO_VOCAB_SIZE,
    ])

    codes_2.extend([
        frame[2] - AUDIO_TOKENS_START - 2 * AUDIO_VOCAB_SIZE,
        frame[3] - AUDIO_TOKENS_START - 3 * AUDIO_VOCAB_SIZE,
        frame[5] - AUDIO_TOKENS_START - 5 * AUDIO_VOCAB_SIZE,
        frame[6] - AUDIO_TOKENS_START - 6 * AUDIO_VOCAB_SIZE,
    ])

print("SNAC level 0 codes:", len(codes_0))
print("SNAC level 1 codes:", len(codes_1))
print("SNAC level 2 codes:", len(codes_2))

# ---------------------------------------------------------
# 3. Decode with the SNAC model already loaded
# ---------------------------------------------------------
print("Decoding with SNAC...")

codes_0 = torch.tensor(
    [codes_0],
    dtype=torch.int32,
    device=snac_codec.device,
)

codes_1 = torch.tensor(
    [codes_1],
    dtype=torch.int32,
    device=snac_codec.device,
)

codes_2 = torch.tensor(
    [codes_2],
    dtype=torch.int32,
    device=snac_codec.device,
)

with torch.inference_mode():
    audio_tensor = snac_codec.model.decode(
        [codes_0, codes_1, codes_2]
    )

audio = audio_tensor.squeeze().detach().float().cpu().numpy()

# ---------------------------------------------------------
# 4. Convert to PCM16
# ---------------------------------------------------------
audio = np.clip(audio, -1.0, 1.0)

pcm16 = (audio * 32767.0).astype(np.int16)

print("Audio samples:", len(pcm16))
print("Duration:", round(len(pcm16) / 24000, 3), "seconds")

# ---------------------------------------------------------
# 5. Save WAV
# ---------------------------------------------------------
OUTPUT_DIR = PROJECT_DIR / "outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

wav_path = OUTPUT_DIR / "svara_test.wav"

with wave.open(str(wav_path), "wb") as wf:
    wf.setnchannels(1)
    wf.setsampwidth(2)
    wf.setframerate(24000)
    wf.writeframes(pcm16.tobytes())

print("Saved:", wav_path)

# ---------------------------------------------------------
# 6. Play it inside the notebook
# ---------------------------------------------------------
display(Audio(str(wav_path)))
```
